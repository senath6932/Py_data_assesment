### Reading and combining data

```python
import pandas as pd
from pathlib import Path

csv_dir = Path("Wine_Stats_csv")
all_dataframes = []

csv_list = sorted(csv_dir.glob("*.csv"))[:8]

for file in csv_list:
    df = pd.read_csv(file, encoding="utf-8", delimiter=",")
    all_dataframes.append(df)

wine_df = pd.concat(all_dataframes, ignore_index=True)

with pd.option_context('display.max_rows', None, 'display.max_columns', None, 'display.width', 1000):
    print(wine_df)
```

### Display the structure of the DataFrame

```python
print(wine_df.info())
```

### Remove duplicate records

```python
before = wine_df.shape[0]
wine_df.drop_duplicates(inplace=True)
print(f"Removed {before - wine_df.shape[0]} duplicate rows")
```

### Remove rows with null values

```python
nulls = wine_df.isnull().sum().sum()
if nulls > 0:
    wine_df.dropna(inplace=True)
    print(f"Removed {nulls} null values")
else:
    print("No null values found")
```
